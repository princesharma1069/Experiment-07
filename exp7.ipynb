```python
# Factorial of a number
n = int(input("Enter n:"))
i=1
if(n==1 or n==0):
  print("Factorial is 1")
else:
  while n>0:
    i *= n
    n -= 1
  print("Factorial is",i)
```

```python
# Reverse a number
n = int(input("Enter number:"))
rev =0
while n>0:
  digit =n%10
  rev =rev*10+digit
  n//=10
print("Reversed number:",rev)
```

```python
#Check Palindrome number
n = int(input("Enter number:"))
temp=n
rev =0
while n>0:
  digit =n%10
  rev =rev*10+digit
  n//=10
if temp==rev:
  print("Palindrome")
else:
  print("Not Palindrome")
```

```python
#Check Palindrome string
s = (input("Enter string: "))
i,j = 0,len(s) -1
is_palindrome = True
while i<j:
  if s[i]!= s[j]:
    is_palindrome =False
    break
  i += 1
  j -= 1
if is_palindrome:
  print("palindrome")
else:
  print("Not palindrome")
```

```python
s = (input("Enter string: "))
rev = s[::-1]
if s==rev:
  print("Palindrome")
else:
  print("Not Palindrome")
```

```python
# count digits in a number
num = int(input("Enter number:"))
count = 0
while num>0:
  count+=1
  num//=10
print("Number of digits:",count)
```

```python
#search an element in the list
num=[10,20,30,40,50]
key =int(input("Enter key:"))
i=0
while i<len(num):
  if num[i]==key:
    print("Element found at index",i)
    break
  i+=1
else:
  print("Element not found")
```
